## Q.1 Write a program to create a NumPy 1D-array with 5 elements and perform basic operations like:
### a) Addition of 2 in all the elements
### b) Multiply 3 with all the elements
### c) Divide every element by 2

## Q.2 Questions on Basic NumPy Array:
### a) Reverse the NumPy array: arr = np.array([1, 2, 3, 6, 4, 5])
### b) Find the most frequent value and their indice(s) in the following arrays:
- i. x = np.array([1,2,3,4,5,1,2,1,1,1])
- ii. y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])

```python
arr = np.array([1, 2, 3, 6, 4, 5])
print("Original Array:", arr)
print("a) Reversed Array:", arr[::-1])
```

```python
print("b) Most frequent value and their indices:\n")

x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])
print("i. x =", x)
values, counts = np.unique(x, return_counts=True)
most_frequent_x = values[np.argmax(counts)]
indices_x = np.where(x == most_frequent_x)[0]
print(f"   Most frequent value: {most_frequent_x} (appears {np.max(counts)} times)")
print(f"   Indices: {indices_x}")

print()

y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])
print("ii. y =", y)
values, counts = np.unique(y, return_counts=True)
most_frequent_y = values[np.argmax(counts)]
indices_y = np.where(y == most_frequent_y)[0]
print(f"   Most frequent value: {most_frequent_y} (appears {np.max(counts)} times)")
print(f"   Indices: {indices_y}")
```

## Q.4 Write program to create a 1-D NumPy array named Aman_Jha with evenly spaced 25 numbers from 10 to 100 using linspace(). Print the dimensions of the array, shape, total elements, the data type of each element and total number of bytes consumed by the array. Find the transpose of this array using reshape() attribute. Can we do the same with T attribute?

```python
Aman_Jha = np.linspace(10, 100, 25)
print("Array Aman_Jha:", Aman_Jha)

print("\nDimensions (ndim):", Aman_Jha.ndim)
print("Shape:", Aman_Jha.shape)
print("Total elements (size):", Aman_Jha.size)
print("Data type:", Aman_Jha.dtype)
print("Total bytes consumed (nbytes):", Aman_Jha.nbytes)

print("\nTranspose using reshape():")
transposed = Aman_Jha.reshape(25, 1)
print(transposed)

print("\nUsing T attribute on 1-D array:")
print(Aman_Jha.T)
print("\nNote: The T attribute on a 1-D array returns the same 1-D array (no effect).")
print("To get a column vector using T, we first need to reshape it to 2-D:")
print(Aman_Jha.reshape(1, 25).T)
```

## Q.5 Create a 2-D Array of three rows and four columns, named ucs420_Aman_Jha with values – 10, 20, 30, 40, 50, 60, 70, 80, 90, 15, 20, 35. Compute the mean, median, max, min, unique elements. Reshape the array to four rows and three columns and name it as reshaped_ucs420_Aman_Jha. Resize the array to two rows and three columns and name it as resized_ucs420_Aman_Jha.

```python
ucs420_Aman_Jha = np.array([10, 20, 30, 40, 50, 60, 70, 80, 90, 15, 20, 35]).reshape(3, 4)
print("ucs420_Aman_Jha (3x4):")
print(ucs420_Aman_Jha)

print("\nMean:", np.mean(ucs420_Aman_Jha))
print("Median:", np.median(ucs420_Aman_Jha))
print("Max:", np.max(ucs420_Aman_Jha))
print("Min:", np.min(ucs420_Aman_Jha))
print("Unique elements:", np.unique(ucs420_Aman_Jha))

reshaped_ucs420_Aman_Jha = ucs420_Aman_Jha.reshape(4, 3)
print("\nreshaped_ucs420_Aman_Jha (4x3):")
print(reshaped_ucs420_Aman_Jha)

resized_ucs420_Aman_Jha = np.resize(ucs420_Aman_Jha, (2, 3))
print("\nresized_ucs420_Aman_Jha (2x3):")
print(resized_ucs420_Aman_Jha)
```
