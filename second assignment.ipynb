```python
def isPrime(n):
    if n < 2:
        return False

    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False

    return True
```

Q4. Using the same 8 digits of your roll number from Q1 (before multiplying by 10), create set A = {digit × 7 for
each digit} and set B = {digit × 9 for each digit}. Print both sets.

vi. Find the union of A and B — all unique values across both sets.

vii. Find the intersection of A and B — values common to both.

viii. Find A − B and B − A separately (not combined) using difference(), and explain in one line how this
differs from symmetric_difference().

ix. Find the symmetric difference of A and B.

x. Check whether A is a subset of B, and whether B is a superset of A, using issubset() and issuperset().

xi. Ask the user to input a value X. Remove X from set A if present using discard() (not remove()) — explain
in one line why discard() is safer than remove() when you're not sure the value exists.

```python

digits = [1,0,2,4,1,7,0,2,1,5]

A = {x * 7 for x in digits}
B = {x * 9 for x in digits}

print("A:", A)
print("B:", B)

union = A.union(B)
print("Union:", union)

intersection = A.intersection(B)
print("Intersection:", intersection)

A_B = A.difference(B)
B_A = B.difference(A)

print("A-B:", A_B)
print("B-A:", B_A)

print("Difference gives one-sided values; symmetric difference gives non-common values.")

symmetric = A.symmetric_difference(B)
print("Symmetric:", symmetric)

print("A subset B:", A.issubset(B))
print("B superset A:", B.issuperset(A))

X = int(input("Enter X: "))

A.discard(X)

print("A after discard:", A)
print("discard() does not give an error if X is absent.")
```
